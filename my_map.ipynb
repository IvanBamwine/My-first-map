```python
import geopandas as gpd

gdf = gpd.read_file('Uganda_Districts_2020.shp')

print("Columns in file:",
 gdf.columns.tolist())
gdf.head()
```

```python
Kabarole = gdf[gdf['ADM2_EN'] ==
               'Kabarole']
Kabarole.head()
```

```python
Kabarole.plot()
```

```python
Kabarole.to_file('Kabarole_District.shp')
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt

subcounties = gpd.read_file('Uganda_Subcounties_2020.shp')
```

```python
print("Columns in shapefile:", subcounties.columns)
```

```python
Kabarole_Subcounties = subcounties[subcounties['ADM2_EN'] == 'Kabarole']
Kabarole_Subcounties.head()
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(10, 10))

Kabarole.plot(ax=ax, color='none' ,
              edgecolor='blue' , linewidth=2,
              label='District Boundary')

Kabarole_Subcounties.plot(ax=ax,
                          color='lightgreen', edgecolor='red',
                          alpha=0.6, label='Subcounties')

plt.title('Map of Kabarole district showing the subcounties')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.legend()
plt.show()
```
