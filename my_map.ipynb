```python
import geopandas as gpd
import os

os.environ['SHAPE_RESTORE_SHX'] = 'YES'

gdf = gpd.read_file('Uganda_Districts_2020.shp')

print("Columns in file:",
 gdf.columns.tolist())
gdf.head()
```

```python
Kabarole = gdf[gdf['ADM2_EN'] ==
               'Kabarole']
Kabarole.head()
```

```python
Kabarole.plot()
```

```python
import geopandas as gpd
import matplotlib.pyplot as plt

subcounties = gpd.read_file('Uganda_Subcounties_2020.shp')
```

```python
print("Columns in shapefile:", subcounties.columns.tolist())
subcounties.head()
```

```python
Kabarole_Subcounties = subcounties[subcounties['ADM2_EN'] == 'Kabarole']
Kabarole_Subcounties.head()
```

```python
schools = gpd.read_file('Uganda_Education_Facilities.shp')
print(schools.columns.tolist())
schools.head()
schools = schools.to_crs(Kabarole.crs)
Kabarole_schools = gpd.clip(schools,Kabarole)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import matplotlib.lines as mlines


Kabarole = gdf[gdf['ADM2_EN'] == 'Kabarole']
Kabarole_Subcounties = subcounties[subcounties['ADM2_EN'] == 'Kabarole']

fig, ax = plt.subplots(figsize=(10, 10))

Kabarole.plot(ax=ax, color='none' ,
              edgecolor='black' , linewidth=2,
              label='District Boundary')

Kabarole_Subcounties.plot(ax=ax,
                          color='white', edgecolor='red',
                          alpha=0.6, label='Subcounties')

Kabarole_schools.plot(ax=ax, color="purple",marker="^", markersize=35, zorder=5)

plt.title('Map of Kabarole district showing the Subcounties and Education facilities')
plt.xlabel('Longitude')
plt.ylabel('Latitude')

district_patch = mpatches.Patch(
    facecolor='none', edgecolor='blue',
    label='District Boundary')

subcounties_patch = mpatches.Patch(
    facecolor='lightgreen', edgecolor='red',
    label='Subcounties')

school_marker = mlines.Line2D([],
                              [],
    color='purple', markersize=7, marker='^', linestyle="None",
    label='Education Facilities')


plt.legend(handles=[district_patch, subcounties_patch, school_marker])
plt.show()
```
